```python
import os
from pdf2image import convert_from_path
import pandas as pd
import shutil

def pdf_to_flashcards(pdf_path, base_output_folder):
    # Obtém o nome do PDF sem a extensão
    pdf_name = os.path.splitext(os.path.basename(pdf_path))[0]

    # Cria uma nova pasta com o nome do PDF
    output_folder = os.path.join(base_output_folder, pdf_name)
    anki_media_folder = os.path.join(output_folder, 'Anki_Media')

    # Cria as pastas necessárias
    os.makedirs(output_folder, exist_ok=True)
    os.makedirs(anki_media_folder, exist_ok=True)

    # Converte o PDF em imagens
    pages = convert_from_path(pdf_path)

    pergunta_imgs = []
    resposta_imgs = []

    for page_num in range(len(pages)):
        img_name = f'{pdf_name}_page_{page_num + 1}.jpg'  # Adiciona o nome do PDF ao nome da imagem
        img_path = os.path.join(anki_media_folder, img_name)  # Salva as imagens diretamente em 'Anki_Media'
        pages[page_num].save(img_path, 'JPEG')

        # Adiciona o caminho da imagem em formato HTML
        img_html = f'<img src="{img_name}">'
        if page_num % 2 == 0:  # Páginas ímpares (1, 3, 5...) são perguntas
            pergunta_imgs.append(img_html)
        else:  # Páginas pares (2, 4, 6...) são respostas
            resposta_imgs.append(img_html)

    # Cria um DataFrame com os nomes das imagens em formato HTML
    df = pd.DataFrame({'Pergunta': pergunta_imgs, 'Resposta': resposta_imgs})

    # Salva o DataFrame como um arquivo CSV na nova pasta
    df.to_csv(os.path.join(output_folder, 'flashcards.csv'), index=False)

def processar_pasta_pdfs(pasta_pdfs, base_output_folder):
    # Percorre todos os arquivos da pasta e processa os PDFs
    for pdf_file in os.listdir(pasta_pdfs):
        pdf_path = os.path.join(pasta_pdfs, pdf_file)
        if pdf_file.lower().endswith('.pdf'):  # Verifica se é um arquivo PDF
            print(f'Processando {pdf_file}...')
            pdf_to_flashcards(pdf_path, base_output_folder)

# Defina a pasta que contém os PDFs
pasta_pdfs = '/content/drive/MyDrive/A PDF_to_Flashcards'  # Substitua pelo caminho correto da sua pasta de PDFs
base_output_folder = '/content/drive/MyDrive/Flashcards para exportar'  # Pasta base para salvar as imagens e CSVs

# Chama a função para processar todos os PDFs da pasta
processar_pasta_pdfs(pasta_pdfs, base_output_folder)

```
